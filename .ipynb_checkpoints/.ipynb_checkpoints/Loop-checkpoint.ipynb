```python
#A loop in Python is a control structure that repeatedly executes a block of code until a specified condition is met or all items in a sequence have been processed.
```

# For Loop


```python
# For loop is used to iterate over a sequence (like a list, tuple or rnage) and execute the block of code once for each item in sequence
```
