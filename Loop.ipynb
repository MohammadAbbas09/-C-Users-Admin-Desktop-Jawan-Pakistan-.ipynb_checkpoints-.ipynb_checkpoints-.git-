# Nested Loop

```python
week_days= ['Monday', 'Tuesday', 'Wednesday', 'Thrusday', 'Friday']
time_slots= ['9AM-10AM', '10AM-11AM', '11AM-12AM']
```

```python
for day in week_days:
    print(day)
```

```python
for day in week_days:
    print(day)
    for slot in time_slots:
        print(slot)
    print('--------')
```

```python
for day in week_days:
    print(day)
    for slot in time_slots:
        print(day,slot)
    print('--------')
```
